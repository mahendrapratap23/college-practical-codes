# Practical 6: Map Coloring Using Backtracking

**Objective:** Solve a constraint satisfaction problem using **backtracking** and **consistency checking**.

In this practical, we will solve one very small **Map Coloring Problem**.

The rule is simple:

> Two connected regions cannot have the same color.

We will use only three colors: **Red, Green, and Blue**.

## 1. What Is a Constraint Satisfaction Problem?

A **Constraint Satisfaction Problem (CSP)** has three parts:

- **Variables**: the things we need to decide.
- **Values**: the choices available.
- **Constraints**: the rules we must follow.

For our map:

- Variables = regions `A`, `B`, `C`, and `D`
- Values = `Red`, `Green`, and `Blue`
- Constraint = connected regions must have different colors.


## 2. Create the Map

We will use this small map:

```text
A ----- B
|      /|
|    /  |
|  /    |
C ----- D
```

The connections are:

* A is connected to B and C
* B is connected to A, C, and D
* C is connected to A, B, and D
* D is connected to B and C

The connected regions are stored in a Python dictionary.


````python
map_data = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

colors = ['Red', 'Green', 'Blue']
````

### Understanding the map code

For example:

```python
'A': ['B', 'C']
```

means region `A` touches regions `B` and `C`.

So `A` cannot have the same color as `B` or `C`.

The list:

```python
colors = ['Red', 'Green', 'Blue']
```

contains the colors we are allowed to use.


## 3. Consistency Checking

Before giving a region a color, we check whether that color breaks any rule.

This is called **consistency checking**.

A color is consistent if none of the already-colored neighbours has the same color.


````python
def is_consistent(region, color, assignment):
    for neighbor in map_data[region]:
        if neighbor in assignment and assignment[neighbor] == color:
            return False

    return True
````

### Understanding `is_consistent()`

The function checks every neighbour of the region.

```python
if neighbor in assignment and assignment[neighbor] == color:
    return False
```

means:

> If a neighbour already has the same color, this choice is not allowed.

If no rule is broken, the function returns `True`.


## 4. Backtracking

Backtracking tries one choice at a time.

The steps are:

1. Choose an uncolored region.
2. Try a color.
3. Check whether the color is consistent.
4. If it works, move to the next region.
5. If we get stuck, remove the last choice and try another color.

That last step is called **backtracking**.


````python
def solve_map(assignment):
    # If every region has a color, we are done
    if len(assignment) == len(map_data):
        return assignment

    # Choose the next uncolored region
    for region in map_data:
        if region not in assignment:
            break

    # Try each color
    for color in colors:
        if is_consistent(region, color, assignment):
            assignment[region] = color

            result = solve_map(assignment)

            if result is not None:
                return result

            # Backtrack
            del assignment[region]

    return None
````

## 5. Understanding the Backtracking Code

This part checks whether every region has been colored:

```python
if len(assignment) == len(map_data):
    return assignment
```

Then the program finds one uncolored region and tries each color:

```python
for color in colors:
```

Before using a color, it checks:

```python
is_consistent(region, color, assignment)
```

If a later region cannot be colored, this line removes the last choice:

```python
del assignment[region]
```

Then Python tries another color.

> **Remember:** Try a choice. If it causes a problem later, undo it and try another choice.


## 6. Solve the Map


````python
solution = solve_map({})

print("Solution:")

for region, color in solution.items():
    print(region, "=", color)
````

### One Possible Output

```text
Solution:
A = Red
B = Green
C = Blue
D = Red
```

No two connected regions have the same color, so all constraints are satisfied.


## 7. Quick Recap

- `A`, `B`, `C`, and `D` are the **variables**.
- Red, Green, and Blue are the possible **values**.
- Connected regions having different colors is the **constraint**.
- `is_consistent()` checks whether a choice follows the rule.
- `solve_map()` uses **backtracking** to try choices and undo a choice when needed.

> **Main idea:** Backtracking tries possible answers, while consistency checking stops choices that break the rules.
