````python
class Student:
    def set_data(self, name, age):
        self.name = name
        self.age = age
    def show_data(self):
        print("Name:", self.name)
        print("Age:", self.age)
s1 = Student()
s1.set_data("Mahendra", 20)
s1.show_data()
````

````python
class Rectangle:
    def set_values(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimtr(self):
        return 2 * (self.length + self.width)
o1 = Rectangle()
o1.set_values(7, 3)
print("Area:", o1.area())
print("Perimeter:", o1.perimtr())
````

````python
class Bird:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city
    def chirp(self):
        print(f"{self.name} is chirping. Age: {self.age} and the city is {self.city}")
b1 = Bird("dove", 2, "Mumbai")
b2 = Bird("hummingbird", 4, "Delhi")
b1.chirp()
b2.chirp()
````

````python
class Circle:
    def set_radius(self, radius):
        self.radius = radius
        self.pi = 3.14

    def area(self):
        return self.pi * self.radius**2

    def diameter(self):
        return 2 * self.radius

    def circumference(self):
        return 2 * self.pi * self.radius

c1 = Circle()
c1.set_radius(6)
print("Area:", c1.area())
print("Diameter:", c1.diameter())
print("Circumference:", c1.circumference())
````
